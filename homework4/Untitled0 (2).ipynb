```python
#34
male_sur=np.mean(ages[(sex=="male")&(survived == 1)])
male_notsur=np.mean(ages[(sex=="male")&(survived==0)])
female_sur=np.mean(ages[(sex=="female")&(survived==1)])
female_notsur=np.mean(ages[(sex =="female")&(survived==0)])
print(male_sur,male_notsur,female_sur,female_notsur)
```

Контрольный вопрос: И означает выполнение всех условий одновременно, ИЛИ — выполнение хотя бы одного.

```python
#37
titanic_statistics = df.groupby(
    ["Pclass", "Sex"],
    as_index=False
).agg(
    PassengerCount=("PassengerId", "size"),
    MeanAge=("Age", "mean"),
    SurvivalPercent=("Survived", lambda x: x.mean() * 100),
    MeanFare=("Fare", "mean")
)

titanic_statistics
```

```python
#38
titanic_statistic = df.groupby(
    ['Embarked'],
    as_index=False,
).agg(
    PassengerCount=("PassengerId", "size"),
    SurvivalPercent=("Survived", lambda x: x.mean() * 100),
)
titanic_statistic
```
